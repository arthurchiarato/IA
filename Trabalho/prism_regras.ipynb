```python
import pandas as pd
```

```python
class PrismModel:

    def predict(self, X):
        resultados = []

        for _, row in X.iterrows():

            if row["tear"] == "reduced":
                resultados.append("none")

            elif (row["age"] == "presbyopic" and
                  row["tear"] == "normal" and
                  row["diagnosis"] == "myope" and
                  row["astigmatism"] == "no"):
                resultados.append("none")

            elif (row["diagnosis"] == "hypermetrope" and
                  row["astigmatism"] == "yes" and
                  row["age"] == "pre-presbyopic"):
                resultados.append("none")

            elif (row["age"] == "presbyopic" and
                  row["diagnosis"] == "hypermetrope" and
                  row["astigmatism"] == "yes"):
                resultados.append("none")


            elif (row["astigmatism"] == "no" and
                  row["tear"] == "normal" and
                  row["diagnosis"] == "hypermetrope"):
                resultados.append("soft")

            elif (row["astigmatism"] == "no" and
                  row["tear"] == "normal" and
                  row["age"] == "young"):
                resultados.append("soft")

            elif (row["age"] == "pre-presbyopic" and
                  row["astigmatism"] == "no" and
                  row["tear"] == "normal"):
                resultados.append("soft")

            else:
                resultados.append("none")

        return resultados
```

```python
model = PrismModel()
```

```python
entrada = pd.DataFrame(
    [["young","hypermetrope","no","reduced"]],
    columns=["age","diagnosis","astigmatism","tear"]
)
```

```python
pred = model.predict(entrada)
```

```python
print("Classe prevista:", pred[0])
```
