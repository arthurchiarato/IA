```python
import pandas as pd
import numpy as np
```

```python
data = [
["jovem","miopia","não","reduzida","nenhuma"],
["jovem","miopia","sim","normal","gelatinosa"],
["jovem","hipermetropia","não","normal","gelatinosa"],
["jovem","hipermetropia","sim","normal","dura"],

["pré-presbiopia","miopia","não","reduzida","gelatinosa"],
["pré-presbiopia","miopia","sim","reduzida","nenhuma"],
["pré-presbiopia","miopia","não","normal","dura"],
["pré-presbiopia","hipermetropia","não","reduzida","gelatinosa"],
["pré-presbiopia","hipermetropia","sim","normal","dura"],

["presbiopia","miopia","não","normal","gelatinosa"],
["presbiopia","miopia","sim","normal","dura"],
["presbiopia","miopia","sim","normal","gelatinosa"],
["presbiopia","hipermetropia","não","reduzida","nenhuma"],
["presbiopia","hipermetropia","sim","normal","gelatinosa"],
["presbiopia","hipermetropia","não","normal","gelatinosa"]
]

df = pd.DataFrame(data, columns=["idade","diagnostico","astigmatismo","taxa","lente"])

df
```

```python
def entropia(coluna):
    valores = coluna.value_counts(normalize=True)
    return -sum(valores * np.log2(valores))
```

```python
H_total = entropia(df["lente"])

print("Entropia total:", round(H_total,2))
```

```python
for valor in df["taxa"].unique():
    subset = df[df["taxa"] == valor]
    H = entropia(subset["lente"])
    print(f"Taxa = {valor} → Entropia = {round(H,2)} | Total = {len(subset)}")
```

```python
ganho = H_total

for valor in df["taxa"].unique():
    subset = df[df["taxa"] == valor]
    peso = len(subset) / len(df)
    ganho -= peso * entropia(subset["lente"])

print("Ganho (Taxa Lacrimal):", round(ganho,2))
```

```python
subset_reduzida = df[df["taxa"] == "reduzida"]

print(subset_reduzida["lente"].value_counts())
print("Entropia:", round(entropia(subset_reduzida["lente"]),2))
```

```python
atributos = ["idade", "diagnostico", "astigmatismo"]

for atributo in atributos:
    ganho = entropia(subset_reduzida["lente"])

    for valor in subset_reduzida[atributo].unique():
        sub = subset_reduzida[subset_reduzida[atributo] == valor]
        peso = len(sub) / len(subset_reduzida)
        ganho -= peso * entropia(sub["lente"])

    print(f"Ganho ({atributo}) =", round(ganho,2))
```
